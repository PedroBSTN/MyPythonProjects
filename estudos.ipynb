```python
#listas
def new_func():
    atividades = ['acordar', 'comer', 'estudar', 'arrumar casa', 'fazer exercicios']
    return atividades

atividades = new_func()
print(atividades)

#adicionar coisas na lista
atividades.append('dormir')
print(atividades)

#remover coisas na lista
atividades.remove('estudar')
print(atividades)
# ou 
atividades.pop(1)
print(atividades)
```

```python
#teste
atnv = input('o quer adicionar na rotina')
if atnv == 'nada':
    pass
else:
    atividades.append(atnv)
print(atividades)

ati = input('o que quer remover')

if ati in atividades:
    atividades.remove(ati)
    print(atividades)
elif ati == 'nada':
    pass
else:
    print('n tem essa ativiadade')
print("\n"join(atnv))
```

### Maior e menor valor

```python
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = [25000, 29000, 22200, 17750, 15870, 19900]
vendas_2sem = [19850, 20120, 17540, 15555, 49051, 9650]

vendas_ano = vendas_1sem + vendas_2sem
vendas_maxima = max(vendas_ano)
vendas_minimas = min(vendas_ano)

print(' o maior valor de vendas foi {} e o menor foi {}'.format(vendas_maxima,vendas_minimas))
```
